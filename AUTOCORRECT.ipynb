```python
from textblob import TextBlob

test_data = {
    "I havv a dreem": "I have a dream",
    "I am lernng pythn": "I am learning python",
    "This is goood": "This is good",
    "Artifical intelgence is amazing": "Artificial intelligence is amazing"
}

correct_predictions = 0
total_predictions = len(test_data)

for wrong, expected in test_data.items():
    predicted = str(TextBlob(wrong).correct())

    print("Input     :", wrong)
    print("Expected  :", expected)
    print("Predicted :", predicted)
    print()

    if predicted.lower() == expected.lower():
        correct_predictions += 1

accuracy = (correct_predictions / total_predictions) * 100

print("Accuracy =", accuracy, "%")
```

```python
import pandas as pd

# Load dataset
dataset = pd.read_csv("autocorrect_dataset.csv")

# Create dictionary
word_dict = dict(zip(dataset["Wrong_Word"],
                     dataset["Correct_Word"]))

def autocorrect_sentence(sentence):
    sentence = sentence.lower().strip()

    words = sentence.split()

    corrected_words = []

    for word in words:
        corrected_words.append(
            word_dict.get(word, word)
        )

    return " ".join(corrected_words)


text = input("Enter sentence: ")

result = autocorrect_sentence(text)

print("\nOriginal :", text)
print("Corrected:", result)
```

```python
# ============================================
# MODULE 5: ACCURACY EVALUATION
# ============================================

from textblob import TextBlob

# Test Dataset
test_data = {
    "I havv a dreem": "I have a dream",
    "I am lernng pythn": "I am learning python",
    "This is goood": "This is good",
    "Artifical intelgence is amazing": "Artificial intelligence is amazing"
}

correct_predictions = 0
total_predictions = len(test_data)

for wrong_text, expected_text in test_data.items():

    predicted_text = str(TextBlob(wrong_text).correct())

    print("Input      :", wrong_text)
    print("Expected   :", expected_text)
    print("Predicted  :", predicted_text)
    print("--------------------------------")

    if predicted_text.lower() == expected_text.lower():
        correct_predictions += 1

accuracy = (correct_predictions / total_predictions) * 100

print("\n===== FINAL RESULT =====")
print("Total Test Cases :", total_predictions)
print("Correct Predictions :", correct_predictions)
print("Accuracy : {:.2f}%".format(accuracy))
```
